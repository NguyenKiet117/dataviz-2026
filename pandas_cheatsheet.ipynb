````python
import pandas as pd

data = {
    'Class': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Score': [85, 92, 78, 88, 95, 73],
    'Gender': ['F', 'M', 'F', 'M', 'F', 'M']
}
df = pd.DataFrame(data)

result = df.groupby('Class')['Score'].agg(['mean', 'max', 'count'])
print(result)
````

### 1. groupby()
**Dùng khi nào?**  
Khi muốn tính toán theo nhóm (theo lớp, theo giới tính, theo tháng...).

**Syntax cơ bản:**
```python
df.groupby('column')['target'].agg(['mean', 'sum', 'count'])
df.groupby(['col1', 'col2']).mean()

````python
customers = pd.DataFrame({
    'CustomerID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
})

orders = pd.DataFrame({
    'OrderID': [101, 102, 103, 104],
    'CustomerID': [1, 2, 1, 5],
    'Amount': [250, 150, 300, 200]
})

merged = pd.merge(customers, orders, on='CustomerID', how='left')
print(merged)
````

### 2. merge()
**Dùng khi nào?**  
Khi muốn ghép 2 bảng theo cột chung (giống SQL JOIN).

**Syntax:**
```python
pd.merge(df1, df2, on='key', how='left/right/inner/outer')

````python
df = pd.DataFrame({
    'Region': ['North', 'North', 'South', 'South'],
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 80],
    'Profit': [20, 30, 50, 15]
})

pivot = pd.pivot_table(df, values='Sales', index='Region', columns='Product', aggfunc='sum')
print(pivot)
````

### 3. pivot_table()
**Dùng khi nào?**  
Tạo bảng tổng hợp chéo (cross-tabulation), dễ nhìn như Excel PivotTable.

**Syntax:**
```python
pd.pivot_table(df, values='value', index='row', columns='col', aggfunc='sum')

````python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Math': [85, 70, 92],
    'English': [78, 88, 95]
})

df['Total'] = df[['Math', 'English']].apply(sum, axis=1)
df['Passed'] = df['Total'].apply(lambda x: 'Yes' if x >= 160 else 'No')
print(df)
````

### 4. apply()

Áp dụng hàm lên từng hàng/cột hoặc từng giá trị.

**Syntax:**
```python

````python
wide = pd.DataFrame({
    'Student': ['Anna', 'Ben'],
    'Math': [90, 75],
    'Science': [85, 88],
    'History': [78, 92]
})

long = pd.melt(wide, id_vars=['Student'], 
               value_vars=['Math', 'Science', 'History'],
               var_name='Subject', value_name='Score')
print(long)
````

````python
# value_counts
import pandas as pd

data = {
    'Class': ['10A', '10A', '10B', '10B', '10A', '10C', '10B'],
    'Gender': ['F', 'M', 'F', 'M', 'F', 'M', 'F'],
    'Score': [85, 92, 78, 88, 95, 73, 80]
}
df = pd.DataFrame(data)

print("Đếm số lượng theo lớp:")
print(df['Class'].value_counts())

print("\nTỷ lệ phần trăm theo lớp:")
print(df['Class'].value_counts(normalize=True).round(3) * 100)

print("\nĐếm theo giới tính:")
print(df['Gender'].value_counts())
````

````python
### value_counts()
**Dùng khi nào?**  
Dùng để đếm tần suất xuất hiện của các giá trị trong một cột (rất hay dùng với categorical data).

**Syntax:**
```python
df['column'].value_counts()
df['column'].value_counts(normalize=True)   # ra tỷ lệ %
````

````python
# crosstab()
pd.crosstab(df['Class'], df['Gender'], margins=True)
````

````python
# fillna()
import pandas as pd
import numpy as np

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, np.nan, 30, np.nan, 28],
    'Score': [85, 72, np.nan, 88, 95],
    'City': ['Hanoi', 'Hanoi', np.nan, 'Da Nang', 'Hanoi']
}
df = pd.DataFrame(data)

print("Trước khi fillna:")
print(df)
print("\nSố missing values:")
print(df.isnull().sum())

# Fillna với các cách khác nhau
df_clean = df.copy()
df_clean['Age'] = df_clean['Age'].fillna(df_clean['Age'].median())
df_clean['Score'] = df_clean['Score'].fillna(df_clean['Score'].mean().round(1))
df_clean['City'] = df_clean['City'].fillna('Unknown')

print("\nSau khi fillna:")
print(df_clean)
````

### fillna()
**Dùng khi nào?**  
Dùng để xử lý missing values (NaN) bằng cách điền giá trị thay thế.

**Các cách điền phổ biến:**
- `df['col'].fillna(value)` → điền 1 giá trị cố định
- `df['col'].fillna(df['col'].median())` → điền median
- `df['col'].fillna(df['col'].mean())` → điền mean
- `df['col'].fillna('Unknown')` → điền chuỗi

**Ví dụ trên:** Điền median cho Age, mean cho Score, và 'Unknown' cho City.

## PHẦN B — KHÁM PHÁ CẤU TRÚC DỮ LIỆU

```python
import seaborn as sns
import pandas as pd

# Load dataset
titanic = sns.load_dataset('titanic')

titanic.head()
```

### head()
5 dòng đầu tiên cho thấy dữ liệu bao gồm thông tin về hành khách: survived, pclass, sex, age, fare, embarked...

Dữ liệu có cả biến số (numeric) và biến phân loại (categorical). 
Cột `deck` và `age` có thể chứa missing values.

```python
titanic.tail()
```

### tail()
5 dòng cuối cùng cho thấy cấu trúc dữ liệu nhất quán. 
Một số hành khách có thông tin age và deck bị thiếu.

```python
print("Shape:", titanic.shape)
titanic.info()
```

### shape và info()
Dataset có **891 hàng** (hành khách) và **15 cột**.

Có 2 kiểu dữ liệu chính: số (int64, float64) và phân loại (object, category, bool). 
Cột `age`, `deck`, `embarked` có missing values.

```python
titanic.describe()
```

### describe()
- `survived`: Tỷ lệ sống sót trung bình khoảng 38.4%.
- `age`: Tuổi trung bình ~29.7, cao nhất 80.
- `fare`: Giá vé trung bình ~32.2, nhưng có sự chênh lệch rất lớn (max 512.33).

```python
# Missing values
missing = titanic.isnull().sum()
missing_percent = (titanic.isnull().sum() / len(titanic) * 100).round(2)

missing_df = pd.DataFrame({
    'Số lượng missing': missing,
    'Tỷ lệ (%)': missing_percent
})
missing_df[missing_df['Số lượng missing'] > 0]
```

### Phân tích Missing Values

- **deck**: 688 missing (77.2%) → Thiếu rất nhiều, có thể bỏ cột này.
- **age**: 177 missing (19.9%) → Tương đối nhiều, cần xử lý (impute median hoặc nhóm theo pclass/sex).
- **embarked**: 2 missing (0.22%) → Rất ít, có thể dropna hoặc điền mode.
- **embark_town**: 2 missing (giống embarked).

**Cột có vấn đề lớn nhất:** `deck` (thiếu quá nhiều nên khó sử dụng).

```python
print("Sex:\n", titanic['sex'].value_counts())
print("\nPclass:\n", titanic['pclass'].value_counts())
print("\nEmbarked:\n", titanic['embarked'].value_counts(dropna=False))
```

### Phân tích 3 cột Categorical

- **sex**: Nam (577) chiếm đa số so với Nữ (314) → Tỷ lệ nam/nữ ≈ 65/35.
- **pclass**: Hạng 3 (491) chiếm nhiều nhất, sau đó là hạng 1 (216) và hạng 2 (184). Hành khách nghèo chiếm ưu thế.
- **embarked**: S (644) nhiều nhất (Southampton), C (168), Q (77). Có 2 giá trị NaN.

```python
print("AGE:")
print("Mean   :", titanic['age'].mean().round(2))
print("Median :", titanic['age'].median())
print("Min    :", titanic['age'].min())
print("Max    :", titanic['age'].max())

print("\nFARE:")
print("Mean   :", titanic['fare'].mean().round(2))
print("Median :", titanic['fare'].median())
print("Min    :", titanic['fare'].min())
print("Max    :", titanic['fare'].max())

print("\nSIBSP:")
print("Mean   :", titanic['sibsp'].mean().round(2))
print("Median :", titanic['sibsp'].median())
print("Min    :", titanic['sibsp'].min())
print("Max    :", titanic['sibsp'].max())
```

### Phân tích 3 cột Numeric

- **age**: Mean ≈ 29.7, Median = 28. Độ tuổi phân bố khá rộng, có trẻ sơ sinh (0.42) và người già (80).
- **fare**: Mean ≈ 32.2, Median = 14.45. Phân bố lệch phải mạnh (max 512), cho thấy một số vé rất đắt.
- **sibsp**: Mean ≈ 0.54, Median = 0. Hầu hết hành khách đi một mình (không có anh em/vợ chồng).

**Điểm bất thường**: Giá vé có outlier rất lớn (512), fare phân bố không đồng đều.

## PHẦN C — PHÂN TÍCH CƠ BẢN

```python
import seaborn as sns
import pandas as pd

titanic = sns.load_dataset('titanic')

# Tỷ lệ sống sót toàn tàu
survival_rate = titanic['survived'].mean() * 100
total_survived = titanic['survived'].sum()
total_passengers = len(titanic)

print(f"Tổng số hành khách: {total_passengers}")
print(f"Số người sống sót: {total_survived}")
print(f"Tỷ lệ sống sót toàn tàu: {survival_rate:.2f}%")
```

### Tỷ lệ sống sót toàn tàu

**Kết quả:** Trong tổng số **891** hành khách, chỉ có khoảng **342** người sống sót, tương đương **38.38%**.

**Giải thích:**  
Con số này cho thấy thảm họa Titanic rất nghiêm trọng — hơn **61%** hành khách đã thiệt mạng. Đây là lý do tại sao "phụ nữ và trẻ em trước" là nguyên tắc nổi tiếng trong sự kiện này.

```python
# Tỷ lệ sống sót theo giới tính
survival_by_sex = titanic.groupby('sex')['survived'].mean() * 100
print(survival_by_sex)
```

### Tỷ lệ sống sót theo Giới tính

- **Nữ**: 74.04% sống sót  
- **Nam**: 18.89% sống sót  

**Nhận xét:** Phụ nữ có tỷ lệ sống sót cao gấp gần **4 lần** nam giới. Điều này phản ánh nguyên tắc "phụ nữ và trẻ em trước" đã được thực thi trên tàu Titanic.

```python
# Tỷ lệ sống sót theo hạng vé
survival_by_pclass = titanic.groupby('pclass')['survived'].mean() * 100
print(survival_by_pclass)
```

### Tỷ lệ sống sót theo Hạng vé (Pclass)

- **Hạng 1**: 62.96%  
- **Hạng 2**: 47.28%  
- **Hạng 3**: 24.24%  

**Nhận xét:** Hành khách càng mua vé đắt (hạng cao) thì càng có tỷ lệ sống sót cao hơn rõ rệt. Hạng 1 có tỷ lệ sống sót gấp **2.6 lần** so với hạng 3.

```python
# Câu 1: Có bao nhiêu hành khách?
print("Tổng số hành khách:", len(titanic))

# Câu 3: Phụ nữ có tỷ lệ sống sót cao hơn nam không?
print("\nTỷ lệ sống sót theo giới tính:")
print(titanic.groupby('sex')['survived'].mean() * 100)

# Câu 7: Vé càng đắt càng dễ sống sót?
print("\nTỷ lệ sống sót theo hạng vé:")
print(titanic.groupby('pclass')['survived'].mean() * 100)
```
